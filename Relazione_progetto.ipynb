# LLM-as-a-Judge

```python
import pandas as pd
```

```python
df = pd.read_csv("./risultati/AETERNA/LLM-as-a-judge/gen_GPT4.1_con_judge.csv")
df.info()
```

```python
df.head()
```

```python
for row in df.itertuples():
    if row.judge_score < 0.5:
        print(f"{row.Index}. {row.question},\nDifficulty: {row.difficulty} - Judge score: {row.judge_score}")
```

```python
METRICS = [
    "answer_correctness",
    "semantic_similarity",
    "faithfulness",
    "context_recall",
    "context_precision",
    "answer_relevancy"
]
```

```python
df = pd.read_csv("./risultati/AETERNA/LLM-as-a-judge/eval_GPT4.1_con_judge.csv")
df.info()
```

```python
insufficienti = df["judge_score"] < 0.5

df.loc[insufficienti, METRICS] = 0.0
df.loc[insufficienti, "noise_sensitivity(mode=relevant)"] = 1.0
        
df.describe()
```
